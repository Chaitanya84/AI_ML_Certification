**DATA VISUALIZATION**

Scatter Plot:

Used to visualize relationships between two continuous variables. Helps in identifying trends, clusters, and outliers.

Syntax:
plot(x, y, type, col, main, xlab, ylab, pch, cex)

where

**x**:	A numeric vector (for x-axis values), **y**:	A numeric vector (for y-axis values)

**type**:	Type of plot to be drawn: "p" for points, "l" for lines, "b" for both, etc.

**col**:	Color of the points or lines (e.g., "red", "blue", 2 for red, etc.)

**main**:	Main title of the plot

**xlab**:	Label for the x-axis, **ylab**:	Label for the y-axis

**pch**:	Plotting character (symbol used for points, e.g., 1-25)

**cex**:	Size of points (default is 1. Bigger number = bigger point)

```r
# Example:
x <- rnorm(100, mean = 50, sd = 10)  # Random normal data
y <- x + rnorm(100, mean = 0, sd = 5)  # Adding some noise

plot(x, y,
     main = "Scatter Plot",
     xlab = "X Variable",
     ylab = "Y Variable",
     col = "blue",
     pch = 16,
     cex = 1.2)

# Interpretations:
# The points show the relationship between x and y.
# If there's an upward trend, it suggests a positive correlation;
# if downward, a negative correlation.
```

**LINE PLOT**

1. Shows trends over time or sequential data.
2. Useful for time series and tracking changes.

Syntax:

plot(x, y, type = "l")

lines(x, y, col, lwd)

where

**Line width**: (default is 1. Larger numbers = thicker lines)

```r
# Example:
x <- 1:50  # Sequence from 1 to 50
y <- cumsum(rnorm(50, mean = 0, sd = 1))  # Cumulative sum of random normal numbers

plot(x, y, type = "o", col = "red",
     main = "Line Plot",
     xlab = "Time", ylab = "Value")

# Interpretations:
# The line connects sequential points, showing a trend over time.
# Useful in stock market trends, temperature variations, etc.
```

**HISTOGRAM**
1. Shows the distribution of a continuous variable.
2. Helps in identifying skewness, peaks, and spread.

Syntax:

hist(x, breaks, col, main, xlab, border)

where

**breaks**:	Number of bins (bars) to divide the data into (can also accept a vector of break points)

```r
# Example:
data <- rnorm(500, mean = 100, sd = 15)

hist(data,
     breaks = 30,
     col = "lightblue",
     main = "Histogram of Data",
     xlab = "Values",
     border = "black")

# Interpretations:
# The shape of the histogram helps determine the distribution
# (normal, skewed, uniform, etc.).
# The number of breaks determines the bin size.
```

**Box Plot**


1. Summarizes the distribution with median, quartiles, and outliers.
2. Helps detect skewness and outliers.


Syntax:

boxplot(x, main, ylab, col, border)

```r
# Example:
data <- rnorm(100, mean = 50, sd = 10)

boxplot(data,
        main = "Boxplot of Data",
        ylab = "Values",
        col = "darkgreen",
        border = "black")

# Interpretations:
# The middle line is the median.
# The box shows the interquartile range (IQR).
# Whiskers show variability outside the IQR;
# points beyond whiskers are outliers.
# lower line of box is Q1 and it tells that 25% data is below and 75% is above
# upper line of box is Q3 and it tells that 75% data is below and 25% is above
# Q2 is median that is middle line
# circles are outlier, that is outside Q1-1.5*IQR and Q3+1.5*IQR (1.5 is inbuilt and can be changed)
```

**Bar Plot**

Compares frequencies or counts of categorical variables.

Syntax:

barplot(height, names.arg, col, main, ylab)

```r
# Example:
categories <- c("A", "B", "C", "D", "E")
values <- sample(10:50, 5)  # Random frequencies

barplot(values,
        names.arg = categories,
        col = "darkgreen",
        main = "Bar Plot",
        ylab = "Frequency")

# Interpretations:
# Each bar represents the frequency of a category.
# The height shows how much of each category is present.
```

**Pie Chart**

Shows proportions among different categories.

Syntax:

pie(x, labels, col, main)

```r
# Example:
slices <- sample(10:50, 5)  # Random proportions
labels <- c("A", "B", "C", "D", "E")

pie(slices,
    labels = labels,
    col = rainbow(5),
    main = "Pie Chart")

# Interpretations:
# The sizes of the slices indicate the proportion of each category.
# Often criticized for being hard to compare exact values.
```

Multivariate analysis with R


**Description**

measure_data[, c("chest", "waist", "hips")]
This part selects specific columns from the measure_data data frame.

c("chest", "waist", "hips") is a vector of column names you want to keep.

measure_data[, ...] tells R to include all rows (before the comma) but only the columns listed after the comma.

```r
# Reading the CSV file
measure_data <- read.csv("measure_data.csv")

# Display the first few rows of the data
head(measure_data)

# Covariance matrix (excluding the gender column)
cov_matrix <- cov(measure_data[, c("chest", "waist", "hips")])
print("Covariance Matrix:")
print(cov_matrix)
```

```r
# Correlation matrix
cor_matrix <- cor(measure_data[, c("chest", "waist", "hips")])
print("Correlation Matrix:")
print(cor_matrix)
```

```r
# Covariance and correlation for male and female separately
# Split data by gender
male_data <- subset(measure_data, gender == "male")[, c("chest", "waist", "hips")]
female_data <- subset(measure_data, gender == "female")[, c("chest", "waist", "hips")]

# Covariance and Correlation for Males
 cat("\nCovariance Matrix - Males:\n")
print(cov(male_data))
 cat("\nCorrelation Matrix - Males:\n")
print(cor(male_data))

# Covariance and Correlation for Females
cat("\nCovariance Matrix - Females:\n")
print(cov(female_data))
cat("\nCorrelation Matrix - Females:\n")
print(cor(female_data))
```
