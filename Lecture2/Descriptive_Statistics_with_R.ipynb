**Measures of Central Tendency**

```r
# Example dataset
x <- c(5, 5,30,35,30, 15, 20, 25, 30, 35, 40)

# Mean
mean(x)

# Median
median(x)

# R does not have a standard in-built function to calculate mode.

# Mode (custom function)
 mode_func <- function(v) {
  uniq_vals <- unique(v)
  uniq_vals[which.max(tabulate(match(v, uniq_vals)))]
 }
 mode_func(x)
```

**Measures of Dispersion**

```r
# Range
range(x)

# Variance
var(x)

# Standard Deviation
sd(x)

```

**Data Distribution and Shape**

```r
# Skewness
 skewness(x)

# Kurtosis
kurtosis(x)
```

**Data Visualization**
